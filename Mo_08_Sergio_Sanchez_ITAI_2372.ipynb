## 3. Proof-of-Concept Simulation: FaunaForge Behavioral Engine
To accompany our pitch deck narrative, the Python prototype below demonstrates how **FaunaForge AI** dynamically shifts ecosystem agent decisions based on real-time world state inputs (player hunting pressure, resource scarcity, and weather), contrasting rigid scripted logic against adaptive generative decision-making.

```python
import json

class FaunaForgeAgent:
    """
    Simulated lightweight decision engine for FaunaForge AI.
    Translates environmental telemetry into adaptive behavioral directives.
    """
    def __init__(self, species: str, base_temperament: str):
        self.species = species
        self.base_temperament = base_temperament
        self.memory_threat_level = 0.0  # 0.0 (safe) to 1.0 (high player hunting pressure)

    def evaluate_world_state(self, player_proximity_m: int, grass_availability: float, recent_losses: int) -> dict:
        # Update persistent herd memory
        if recent_losses > 0:
            self.memory_threat_level = min(1.0, self.memory_threat_level + (0.35 * recent_losses))
        else:
            self.memory_threat_level = max(0.0, self.memory_threat_level - 0.05)

        # Generative behavioral policy selection
        if self.memory_threat_level >= 0.7:
            directive = "MIGRATE_TO_HIGH_COVER_BIOME"
            rationale = "High recent predation detected; abandoning open pasture for dense forest."
            vigilance = "MAXIMUM (Sentinels posted)"
        elif grass_availability < 0.3:
            directive = "DISPERSE_AND_FORAGE_WIDE"
            rationale = "Local pasture depleted; splitting herd into smaller foraging subgroups."
            vigilance = "ELEVATED"
        elif player_proximity_m < 40 and self.base_temperament == "defensive":
            directive = "FORM_DEFENSIVE_PERIMETER"
            rationale = "Close proximity contact; adults shielding juveniles."
            vigilance = "HIGH"
        else:
            directive = "GRAZE_AND_REST"
            rationale = "Optimal forage and low threat memory; conserving energy."
            vigilance = "LOW"

        return {
            "species": self.species,
            "threat_memory_score": round(self.memory_threat_level, 2),
            "selected_directive": directive,
            "vigilance_state": vigilance,
            "ai_rationale": rationale
        }

# Simulate 3 consecutive in-game days
herd = FaunaForgeAgent(species="Highland Bighorn Grazer", base_temperament="defensive")

scenarios = [
    {"day": 1, "desc": "Calm morning, abundant pasture", "prox": 150, "grass": 0.85, "losses": 0},
    {"day": 2, "desc": "Players ambush herd at watering hole", "prox": 25, "grass": 0.70, "losses": 2},
    {"day": 3, "desc": "Players return to same watering hole", "prox": 120, "grass": 0.65, "losses": 0},
]

print("=" * 80)
print("FAUNAFORGE AI — LIVE ECOSYSTEM AGENT TELEMETRY DEMO")
print("=" * 80)

for s in scenarios:
    decision = herd.evaluate_world_state(s["prox"], s["grass"], s["losses"])
    print(f"\n[DAY {s['day']}] Event: {s['desc']}")
    print(json.dumps(decision, indent=2))
```
